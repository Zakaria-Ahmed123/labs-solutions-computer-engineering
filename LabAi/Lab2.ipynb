# Exercise 2 : Matrix

```python
grades = [
    [14,12,16],
    [9,13,11], 
    [17,15,14],
    [8,10,12]
]

print(grades)
```

```python
print(grades[1])
```

```python
L = []
for row in grades:
    sum = 0
    for elem in row:
        sum += elem
    L.append(sum / len(row))
L
```

```python
n_rows = 4
n_cols = 3
sum = [0, 0, 0]
avg = [0, 0, 0]
for student in grades:
    for i in range(n_cols):
        sum[i] = student[i] + sum[i]
for i in range(n_cols):
    avg[i] = sum[i] / n_rows
avg
```

```python
L = []
for row in grades:
    sum = 0
    for elem in row:
        sum += elem
    L.append(sum / len(row))
sum = 0
for i in L:
    sum += i
avg = sum / len(L)
avg
```

```python
grades = [
    [14,12,16],
    [9,13,11], 
    [17,15,14],
    [8,10,12]
]

maximum = grades[0][0]
minimum = grades[0][0]
for row in grades:
    for elem in row:
        if elem > maximum:
            maximum = elem
        if elem < minimum:
            minimum = elem

print(minimum)
print(maximum)          
```

```python
L = []
for row in grades:
    sum = 0
    for elem in row:
        sum += elem
    L.append(sum / len(row))
count = 0 
for i in L:
    if i >= 10:
        count += 1
count
```

```python
L.sort()
L
```
